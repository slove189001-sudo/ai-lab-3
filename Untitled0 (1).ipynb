```python
def vacuum_agent(location, status):
    if location not in ("A", "B"):
        return "Unknown Location"
    if status == "dirty":
        return "Suck"
    if location == "A":
        return "Right"
    return "Left"


print("A dirty:", vacuum_agent("A", "dirty"))
print("A clean:", vacuum_agent("A", "clean"))
print("B dirty:", vacuum_agent("B", "dirty"))
print("B clean:", vacuum_agent("B", "clean"))
```

```python
percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")]
for location, status in percepts:
    print(location, status, "->", vacuum_agent(location, status))
```

```python
chosen_actions = []
for location, status in percepts:
    chosen_actions.append(vacuum_agent(location, status))
print(chosen_actions)
```

```python
suck_count = 0
for action in chosen_actions:
    if action == "Suck":
        suck_count = suck_count + 1
print("Suck occurred", suck_count, "times")
```

```python
visited_locations = set()
for location, status in percepts:
    visited_locations.add(location)
print(visited_locations)
```

```python
print("--- Challenge 5: choose_cheapest ---")
def choose_cheapest(costs):
    low = costs[0]
    for c in costs:
        if c < low:
            low = c
    return low
print(choose_cheapest([14, 3, 9, 21, 6]))
```

```python
print("--- Challenge 7: Travel agent ---")
def travel_agent(location, weather, road_status):
    if road_status == "closed":
        return "Use Alternate Route"
    if weather == "snowy" or weather == "rainy":
        return "Wait"
    return "Continue"
print(travel_agent("Mingora", "sunny", "open"))
print(travel_agent("Mingora", "rainy", "open"))
print(travel_agent("Mingora", "sunny", "closed"))
```

```python
print("--- Challenge 8: Two-room vacuum simulation ---")
def vacuum_agent(location, status):
    if status == "dirty":
        return "Suck"
    if location == "A":
        return "Right"
    return "Left"

room_percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")]
for location, status in room_percepts:
    print("Percept:", location, status, "-> Action:", vacuum_agent(location, status))
```
